```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification
```

```python
model_name = "wonrax/phobert-base-vietnamese-sentiment"

tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=False)
model = AutoModelForSequenceClassification.from_pretrained(model_name)
```

```python
text = "bài hát này nghe lúc dở lúc hay"
print("Câu kiểm tra:", text)
```

```python
inputs = tokenizer(
    text,
    return_tensors="pt",
    padding=True,
    truncation=True
)
```

```python
with torch.no_grad():
    output = model(**inputs)

scores = torch.softmax(output.logits, dim=1)
label_id = torch.argmax(scores).item()
confidence = scores[0][label_id].item()
```

```python
labels = {
    0: "Tiêu cực",
    1: "Tích cực",
    2: "Trung tính"
}

print("Kết quả:", labels[label_id])
print("Độ tin cậy:", round(confidence * 100, 2), "%")
```
